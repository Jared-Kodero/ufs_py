```python
import math
from pathlib import Path

import matplotlib.pyplot as plt
import xarray as xr
```

```python
root_path = "/oscar/data/deeps/shared/gfdl_shield/ufs_py/tests/test_case7/run/HIST"
paths = sorted(list(Path(root_path).glob("*.nc")))
```

```python
n = len(paths)
ncols = 3
nrows = math.ceil(n / ncols)


fig, axes = plt.subplots(nrows, ncols, figsize=(6 * ncols, 5 * nrows))
axes = axes.ravel()
for ax, p in zip(axes, paths):
    ds = xr.open_dataset(p)

    da = ds["soilw1"].isel(time=0)
    da.plot(ax=ax)
    ax.set_title(p.name)
for ax in axes[n:]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```
